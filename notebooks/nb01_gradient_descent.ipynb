# Notebook 01:梯度下降視覺化

最小化 $L(x) = (x-3)^2$,梯度為 $\nabla L(x) = 2(x-3)$。

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)  # 固定隨機種子,確保可重現
```

```python
def run_gd(x0, lr, steps=30):
    """對 L(x)=(x-3)^2 執行梯度下降,回傳每一步的 x。"""
    xs = [x0]
    for _ in range(steps):
        grad = 2 * (xs[-1] - 3)
        xs.append(xs[-1] - lr * grad)
    return np.array(xs)

for lr in (0.05, 0.3, 0.9):
    plt.plot(run_gd(10.0, lr), label=f'lr={lr}')
plt.axhline(3, color='gray', ls='--')
plt.xlabel('step'); plt.ylabel('x'); plt.legend(); plt.show()
```
